# **1) First Unique Character**

```python
def first_unique_char(s):
    counts = {}
    for ch in s:
        if ch in counts:
            counts[ch] += 1
        else:
            counts[ch] = 1
    for ch in s:
        if counts[ch] == 1:
            return ch

for t in ["swiss", "aabbcdde", "aabb", "z", "coding"]:
    print(t, "--->", first_unique_char(t))
```

# **2) numbers between 2000 and 3200 which divisible by 7 and not by 3**

```python
result = []
for n in range(2000, 3201):
    if n % 7 == 0 and n % 3 != 0:
        result.append(str(n))
print(",".join(result))
```

# **3) Join characters in given list to a string and capitalize**

```python
chars = ["C", "O", "D", "I", "N", "G"]
text = ""
for c in chars:
    text += c
print(text.capitalize())
```

# **4) make a dictionary with squares for given "n" natural numbers**

```python
# n = int(input())
n = 10 # for output cell
d = {}
for i in range(1, n + 1):
    d[i] = i * i
print(d)
```
